# Introduction to Electricity Theft

## 1. What is Electricity Theft?

Electricity theft is the unauthorized or illegal use of electrical power without paying for the actual amount of electricity consumed.

It occurs when consumers manipulate the electricity supply or metering system to reduce or avoid electricity bills. Electricity theft is one of the major causes of **Non-Technical Losses (NTLs)** and results in significant financial losses for power distribution companies.

---

## 2. Types of Electricity Theft

### Meter Bypass

Electricity is directly drawn from power lines while bypassing the electricity meter. As a result, the actual energy consumption is not recorded.

### Meter Tampering

The electricity meter is physically or electronically modified to record lower consumption than the actual electricity used.

### Illegal Connections (Hooking)

Consumers connect directly to electricity distribution lines without authorization or a valid electricity connection.

### Billing Fraud

Incorrect billing occurs due to manipulated meter readings or fraudulent activities by consumers or utility employees.

---

## 3. Global Statistics and Economic Impact

Electricity theft is a global issue that causes billions of dollars in financial losses every year.

In many developing countries, electricity theft can account for a significant percentage of total electricity distribution losses. It leads to:

- Reduced revenue for electricity companies
- Increased electricity prices for honest consumers
- Reduced power grid reliability
- Additional operational costs for utilities

## How Smart Grids and Smart Meters Work

### 1. Traditional Meters vs. Smart Meters

#### Traditional Meters
Traditional electricity meters are conventional devices that measure the total amount of electricity consumed by a customer. They usually record only the cumulative energy usage in kilowatt-hours (kWh).

The main limitations of traditional meters are:
- Meter readings are collected manually by utility employees.
- They provide limited information about customer electricity usage patterns.
- They cannot detect electricity theft, tampering, or abnormal consumption behavior in real time.
- Communication between customers and utility companies is not possible.

#### Smart Meters
Smart meters are advanced digital electricity meters that automatically measure and transmit detailed electricity consumption data to the utility company.

Unlike traditional meters, smart meters:
- Automatically send meter readings through communication networks.
- Provide real-time or near real-time electricity usage information.
- Enable remote monitoring and control by utility companies.
- Help detect electricity theft, meter tampering, and unusual consumption patterns.
- Support efficient energy management and smart grid operation.

**Difference Between Traditional and Smart Meters**

| Feature | Traditional Meter | Smart Meter |
|---------|------------------|-------------|
| Meter Reading | Manual reading by staff | Automatic remote reading |
| Data Collection | Only total energy consumption | Detailed time-based consumption data |
| Communication | No communication capability | Two-way communication with utilities |
| Theft Detection | Difficult to detect | Can identify abnormal usage and tampering |
| Monitoring | Limited monitoring | Real-time monitoring |
| Control | No remote control | Supports remote operations |

---

### 2. What Data Smart Meters Collect

Smart meters continuously collect detailed electricity usage information, including:

- **Electricity Consumption (kWh):** Measures the amount of electrical energy used by consumers over time.
- **Voltage and Current:** Records electrical parameters to monitor power quality and system conditions.
- **Power Factor:** Measures how efficiently electrical power is being utilized.
- **Time and Date of Usage:** Captures when electricity consumption occurs, enabling time-based analysis.
- **Peak and Off-Peak Consumption:** Identifies high and low demand periods for better load management.
- **Meter Status and Tamper Events:** Detects abnormal conditions such as meter manipulation, disconnection, or tampering attempts.

This detailed information helps electricity utilities:
- Monitor customer electricity usage accurately.
- Improve power distribution efficiency.
- Detect electricity theft and abnormal consumption patterns.
- Plan demand management strategies.

---

### 3. AMI (Advanced Metering Infrastructure)

Advanced Metering Infrastructure (AMI) is a communication system that connects smart meters with electricity utility companies. It enables automatic data collection, monitoring, and two-way communication between consumers and utilities.

AMI consists of the following components:

#### 1. Smart Meters
Smart meters installed at customer locations measure electricity consumption and collect electrical parameters.

#### 2. Communication Network
The communication network transfers data between smart meters and utility companies. Common communication technologies include:
- Radio Frequency (RF)
- Cellular Networks
- Power Line Communication (PLC)

#### 3. Data Concentrators/Gateways
These devices collect data from multiple smart meters and forward it to the utility control center.

#### 4. Utility Control Center
The control center processes and analyzes collected meter data for monitoring, billing, fault detection, and energy management.

### Role of AMI in Smart Grids

AMI enables two-way communication between customers and utilities. It allows utilities to:

- Collect electricity usage data remotely.
- Monitor grid conditions in real time.
- Detect electricity theft and meter tampering.
- Perform remote meter updates.
- Enable remote disconnection or reconnection when required.
- Improve reliability and efficiency of electricity distribution systems.

The dataset represents a real-world electricity fraud detection problem where theft cases are significantly fewer compared to normal consumers. ---

## Dataset Structure

The SGCC dataset follows a **customer-level time-series structure**.

- Each **row represents one electricity consumer**.
- Each **column represents a feature or daily electricity consumption value**.
- Daily consumption columns represent the electricity usage recorded by smart meters over a period of time.
- The target variable indicates whether the consumer is normal or involved in electricity theft.

| Consumer A | 12.5 | 13.2 | 11.8 | ... | 14.6 |

- Sudden drops in electricity usage.

The **FLAG** column represents the class label for electricity theft detection.

# Data Preprocessing

Before applying machine learning models, the SGCC electricity theft detection dataset was preprocessed to improve data quality and extract meaningful patterns from smart meter readings.

---

## 1. Handling Missing Values (NaN)

The dataset contained a significant number of missing values in daily electricity consumption records.

### Problem:
- Missing readings can occur due to communication failures, meter issues, or data collection problems.
- These missing values can affect statistical calculations and reduce model performance.

### Handling Method:
- Missing values were treated using **time-series interpolation** because electricity consumption follows a continuous pattern over time.
- Linear interpolation was applied to estimate missing consumption values based on neighbouring days.
- Remaining missing values after interpolation were handled using:
  - Forward fill (previous valid value)
  - Backward fill (next valid value)

This approach preserved the original consumption behaviour of consumers while reducing data loss.

---

## 2. Zero Consumption Analysis

Zero consumption values were analyzed separately because zero electricity usage can have different meanings.

### Observations:
- Some consumers showed occasional zero consumption days, which can occur due to genuine reasons such as absence or low usage.
- Some consumers had prolonged periods of zero consumption, which may indicate:
  - Abnormal usage patterns.
  - Meter issues.
  - Possible electricity theft behaviour.

### Handling:

Electricity consumption data contains extreme values due to differences in consumer behaviour.

This ensures that genuine consumer behaviour is maintained while reducing the impact of extreme observations.

---

## Summary

The preprocessing steps ensured that:

- Missing values were handled without losing important time-series information.
- Zero consumption patterns were preserved as meaningful behavioural indicators.
- Outliers were controlled carefully instead of being removed blindly.

These preprocessing steps improved data quality and prepared the dataset for effective electricity theft detection using machine learning models.

# 12. Model Interpretation (SHAP & Feature Importance)

Machine learning models such as XGBoost can achieve high predictive performance, but understanding why a model makes a particular decision is important, especially in real-world applications like electricity theft detection.

Model interpretation was performed using:

1. Feature Importance
2. SHAP (SHapley Additive exPlanations)

These techniques help identify the most influential features responsible for predicting electricity theft.

---

# 1. Feature Importance Analysis (XGBoost)

XGBoost provides built-in feature importance scores that measure the contribution of each feature to the model's decision-making process.

The importance of a feature indicates how frequently and effectively it is used by the decision trees to separate normal consumers from electricity theft consumers.

### Purpose of Feature Importance:

- Identify the most influential consumption behaviour patterns.
- Understand which engineered features contribute most to theft detection.
- Remove unnecessary features and improve model interpretability.

The feature importance analysis highlights important characteristics such as:

- Consumption trends.
- Consumption variability.
- Sudden changes in electricity usage.
- Zero consumption behaviour.
- Missing data patterns.

These features provide valuable information about abnormal consumer behaviour.

---

# 2. SHAP Analysis

SHAP (SHapley Additive exPlanations) is an explainable AI technique based on game theory that explains how each feature contributes to an individual prediction.

Unlike traditional feature importance, SHAP provides:

- Global model interpretation.
- Individual prediction explanations.
- Direction of feature impact.

---

# SHAP Summary Plot

The SHAP summary plot shows:

- Overall importance of each feature.
- Whether a feature increases or decreases the probability of electricity theft.
- Distribution of feature impacts across all consumers.

Interpretation:

- Features at the top have the highest impact on model predictions.
- Positive SHAP values push the prediction towards electricity theft.
- Negative SHAP values push the prediction towards normal consumption.
- The colour distribution represents whether high or low feature values contribute to the prediction.

---

# SHAP Feature Importance (Bar Plot)

The SHAP bar plot ranks features based on their average contribution to the model output.

It provides a global view of:

- Which features are most important.
- How much each feature contributes overall to theft detection.

This helps validate whether the model is learning meaningful consumption behaviour rather than random patterns.

---

# SHAP Individual Prediction Explanation

SHAP waterfall plots were used to explain individual consumer predictions.

For a specific consumer, SHAP shows:

- The base prediction probability.
- Features increasing the probability of theft.
- Features decreasing the probability of theft.

This provides transparency in the model's decision-making process.

Example:

A consumer may be classified as theft because of:

- High consumption variability.
- Sudden drops in usage.
- Long zero-consumption periods.
- Unusual consumption trends.

---

# Importance of Model Interpretation in Electricity Theft Detection

Interpretability is important because electricity utilities need confidence before inspecting a consumer.

SHAP and feature importance help answer:

- Why was this consumer flagged?
- Which consumption patterns indicate possible theft?
- Are predictions based on meaningful behavioural patterns?

Therefore, explainable AI improves trust, transparency, and practical usability of the electricity theft detection system.
